## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Bagaimana pengaruh hari libur (holiday) terhadap rata-rata jumlah penyewaan sepeda, dan apakah terdapat perbedaan signifikan dibandingkan hari biasa selama tahun 2011-2012?

## Visualization & Explanatory Analysis

### Pertanyaan 1: Bagaimana pengaruh hari libur (holiday) terhadap rata-rata jumlah penyewaan sepeda, dan apakah terdapat perbedaan signifikan dibandingkan hari biasa selama tahun 2011-2012?

```python
holiday_df = bikesharing_day_df.groupby("holiday")["cnt"].mean().reset_index()

holiday_df["holiday"] = holiday_df["holiday"].replace({
    0: "Hari Biasa",
    1: "Hari Libur"
})

plt.figure(figsize=(8,5))
sns.barplot(
    data=holiday_df,
    x="holiday",
    y="cnt",
    hue="holiday",
    palette="Set2",
    legend=False
)

for index, row in holiday_df.iterrows():
    plt.text(index, row["cnt"], round(row["cnt"], 0), ha='center')

plt.title("Perbandingan Rata-rata Penyewaan Sepeda Hari Biasa vs Hari Libur (2011-2012)")
plt.xlabel("Kategori Hari")
plt.ylabel("Rata-rata Jumlah Penyewaan")
plt.show()
```

- Berdasarkan visualisasi, rata-rata penyewaan sepeda pada hari biasa lebih tinggi dibandingkan hari libur.
- Hal ini menunjukkan bahwa penggunaan sepeda lebih dominan pada hari kerja.
- Perbedaan ini cukup jelas terlihat pada grafik, sehingga dapat disimpulkan bahwa terdapat perbedaan signifikan antara hari libur dan hari biasa dalam hal jumlah penyewaan sepeda.
- Selanjutnya, akan ditampilkan hubungan antara kecepatan angin dan jumlah penyewaan sepeda.

### Pertanyaan 2: Bagimana hubungan antara kecepatan angin (windspeed) terhadap jumlah penyewaan sepeda, dan apakah terdapat tren penurunan penyewaan pada kondisi angin tinggi selama tahun 2011-2012?

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=bikesharing_day_df,
    x="windspeed",
    y="cnt",
    alpha=0.6
)

sns.regplot(
    data=bikesharing_day_df,
    x="windspeed",
    y="cnt",
    scatter=False,
    color="red"
)

plt.title("Hubungan Kecepatan Angin dengan Penyewaan Sepeda (2011-2012)")
plt.xlabel("Windspeed")
plt.ylabel("Jumlah Penyewaan")
plt.show()
```

- Visualisasi ini digunakan untuk melihat hubungan antara kecepatan angin dan jumlah penyewaan sepeda menggunakan scatter plot.
- Titik-titik pada grafik menunjukkan distribusi data, sedangkan garis merah menunjukkan tren hubungan antar variabel.
- Dari grafik terlihat adanya kecenderungan penurunan jumlah penyewaan saat kecepatan angin meningkat.


**Insight:**
- Berdasarkan visualisasi pertama, terdapat perbedaan rata-rata jumlah penyewaan sepeda antara hari biasa dan hari libur. Penyewaan cenderung lebih tinggi pada hari biasa, yang menunjukkan bahwa sepeda lebih sering digunakan untuk aktivitas rutin seperti bekerja atau sekolah.

- Pada visualisasi kedua, terlihat bahwa kecepatan angin memiliki hubungan negatif terhadap jumlah penyewaan sepeda. Semakin tinggi kecepatan angin, jumlah penyewaan cenderung menurun.

- Seluruh proses pada tahap Visualization Data sudah selesai. Selanjutnya akan dilakukan analisis lanjutan berupa clustering sederhana berdasarkan jumlah penyewaan.

## Analisis Lanjutan (Opsional)

1. Kategorisasi Penyewaan


```python
def categorize_cnt(cnt):
    if cnt < 3000:
        return "Rendah"
    elif cnt < 6000:
        return "Sedang"
    else:
        return "Tinggi"

bikesharing_day_df["kategori_penyewaan"] = bikesharing_day_df["cnt"].apply(categorize_cnt)

bikesharing_day_df["kategori_penyewaan"].value_counts()
```

- Kode ini digunakan untuk mengelompokkan jumlah penyewaan sepeda ke dalam tiga kategori, yaitu rendah, sedang, dan tinggi.
- Output menunjukkan bahwa kategori Sedang memiliki jumlah terbanyak (379 hari), diikuti oleh kategori Tinggi (180 hari) dan Rendah (172 hari).
- Hal ini menunjukkan bahwa mayoritas penyewaan sepeda berada pada tingkat penggunaan sedang.
- Selanjutnya, akan dianalisis hubungan kategori penyewaan dengan kondisi cuaca.

2. Hubungan dengan cuaca

```python
weather_category_df = pd.crosstab(
    bikesharing_day_df["weathersit"],
    bikesharing_day_df["kategori_penyewaan"]
)

weather_category_df
```

- Tabel ini menunjukkan distribusi kategori penyewaan pada setiap kondisi cuaca.
- Terlihat bahwa pada cuaca baik, penyewaan didominasi kategori sedang dan tinggi.
- Sementara pada cuaca buruk, penyewaan lebih banyak berada pada kategori rendah.
- Selanjutnya, hasil ini divisualisasikan agar lebih mudah dipahami.

```python
weather_category_df.plot(kind="bar", figsize=(8,5))

plt.title("Distribusi Kategori Penyewaan Berdasarkan Cuaca")
plt.xlabel("Kondisi Cuaca")
plt.ylabel("Jumlah Hari")
plt.xticks(rotation=0)
plt.show()
```

- Grafik memperjelas bahwa semakin buruk kondisi cuaca, jumlah penyewaan sepeda cenderung menurun.
- Dominasi kategori rendah terlihat jelas pada kondisi cuaca buruk.

```python
weather_category_percent = weather_category_df.div(
    weather_category_df.sum(axis=1), axis=0
) * 100

weather_category_percent
```

- Pada cuaca baik, penyewaan didominasi kategori Sedang (53%) dan Tinggi (30%).
- Pada cuaca sedang, kategori Sedang tetap dominan, namun kategori Tinggi menurun.
- Pada cuaca buruk, hampir seluruh penyewaan berada pada kategori Rendah (90%).

- Penyewaan sepeda umumnya berada pada tingkat sedang saat kondisi cuaca baik.
- Semakin buruk cuaca, penyewaan menurun dan didominasi kategori rendah.
- Hal ini menunjukkan bahwa cuaca memiliki pengaruh signifikan terhadap tingkat penggunaan sepeda.
